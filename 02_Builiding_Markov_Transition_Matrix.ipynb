## What is a Markov Chain?
A Markov chain is a stochastic model that describes a sequence of possible events where the probability of each event depends only on the state attained in the previous event. 
It is characterized by the Markov property, which states that the future state is independent of the past states given the present state.

In the context of DNA sequences, a Markov chain can be used to model the transitions between nucleotides (A, C, G, T) based on their probabilities.

```python
import numpy as np
import pandas as pd
```

```python
def compute_transition_matrix(dna_sequence):
    # Define the nucleotide symbols and their indices
    nucleotide_symbols = ['a', 'c', 'g', 't']
    nucleotide_index = {symbol: idx for idx, symbol in enumerate(nucleotide_symbols)}

    # Initialize a 4x4 matrix to count transitions
    transition_counts = np.zeros((4, 4), dtype=int)

    # Count the transitions between consecutive nucleotides
    for i in range(len(dna_sequence) - 1):
        current_nucleotide = dna_sequence[i]
        next_nucleotide = dna_sequence[i + 1]
        if current_nucleotide in nucleotide_index and next_nucleotide in nucleotide_index:
            transition_counts[nucleotide_index[current_nucleotide], nucleotide_index[next_nucleotide]] += 1

    # Normalize the counts to compute probabilities
    transition_probabilities = transition_counts.astype(float)
    row_sums = transition_probabilities.sum(axis=1)
    for i in range(4):
        if row_sums[i] > 0:
            transition_probabilities[i] /= row_sums[i]

    # Convert the matrix to a DataFrame for better readability
    transition_matrix_df = pd.DataFrame(transition_probabilities, columns=nucleotide_symbols, index=nucleotide_symbols)
    return transition_matrix_df
```
