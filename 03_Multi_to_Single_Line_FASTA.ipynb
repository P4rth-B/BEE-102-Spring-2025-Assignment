# Q.3 Convert Multi-line FASTA to Single-line FASTA
The FASTA file format often uses a fixed-width format for sequences, where each line (except the header) contains a fixed number of characters (e.g., 60).

This can make it difficult to work with sequences of interest. For example, a sequence of 600 characters would span 10 lines.

The goal is to convert this format into a simpler one where each header is followed by its entire sequence on a single line.

### Solution Approach

To achieve this conversion, we will:

1. Read the input FASTA file line by line.
2. Identify header lines (lines starting with '>').
3. Concatenate sequence lines until the next header or the end of the file.
4. Write each header followed by its concatenated sequence to the output file.

This will be implemented using Python's file handling and string operations.

```python
def convert_fasta_to_single_line(input_file, output_file):
    with open(input_file, 'r') as infile, open(output_file, 'w') as outfile:
        # Temporary variable to store the current sequence
        current_sequence = ''
        for line in infile:
            # Strip leading and trailing whitespace
            line = line.strip()

            if line.startswith('>'):
                # Write the previous sequence (if any) before the new header
                if current_sequence:
                    outfile.write(current_sequence + '\n')
                    current_sequence = ''

                outfile.write(line + '\n')
            else:
                current_sequence += line

        # Write the last sequence to the output file (if any)
        if current_sequence:
            outfile.write(current_sequence + '\n')

# Execute the function with input and output file names
convert_fasta_to_single_line('multiline_input.fasta', 'output.fasta')
print("FASTA file has been converted and saved.")
```
